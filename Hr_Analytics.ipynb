## Overall Attrition Rate
Converting Attrition column to numeric (1/0) for calculations. 
Pie chart showing the overall split between employees who stayed 
and employees who left.

```python
df['AttritionFlag'] = df['Attrition'].map({'Yes':1,'No':0})

plt.figure(figsize=(6,6))
att_count=df['Attrition'].value_counts()

att_count.reindex(['No','Yes']).plot(kind='pie', autopct='%1.1f%%',
    colors=['#5DCAA5','#D85A30'], labels=['Stayed','Left'])
plt.title('Overall Attrition Split')
plt.show()
```

## Attrition Rate by Department
Comparing attrition rate across departments to identify which 
department has the highest employee turnover.

```python
plt.figure(figsize=(8,4))
att_Groupby=df.groupby('Department')['AttritionFlag'].mean()

att_dept=att_Groupby.sort_values(ascending=False)

att_dept.plot(kind='bar', color='#EF9F27')

plt.title('Attrition Rate by Department')
plt.ylabel('Attrition Rate')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Attrition Rate by Overtime
Comparing attrition rate between employees who work overtime 
and those who don't.

```python

plt.figure(figsize=(6,4))
sns.barplot(data=df, x='OverTime', y='AttritionFlag',hue='OverTime',legend=False,palette=['#5DCAA5','#D85A30'])
plt.title('Attrition Rate by Overtime')
plt.ylabel('Attrition Rate')
plt.show()
```

## Attrition Rate by Job Satisfaction
Attrition rate across job satisfaction levels (1 = Low, 4 = High) 
to check if satisfaction impacts retention.

```python
plt.figure(figsize=(6,4))
sns.barplot(data=df, x='JobSatisfaction', y='AttritionFlag', color='#7F77DD')
plt.title('Attrition Rate by Job Satisfaction')
plt.show()
```

## Correlation Heatmap
Correlation between key numeric HR factors and attrition. 
Values closer to +1 or -1 indicate stronger relationships 
with employee attrition.

```python
num_cols = ['Age','DailyRate','DistanceFromHome','MonthlyIncome',
    'JobSatisfaction','WorkLifeBalance','YearsAtCompany',
    'YearsSinceLastPromotion','NumCompaniesWorked','AttritionFlag']

plt.figure(figsize=(10,8))
corr = df[num_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', center=0)
plt.title('Correlation Heatmap - HR Factors')
plt.tight_layout()
plt.show()
```

## Salary Distribution: Stayed vs Left
Monthly income distribution compared between employees who 
stayed and employees who left, using a boxplot to show median, 
quartiles and outliers.

```python
plt.figure(figsize=(8,4))
sns.barplot(data=df, x='Attrition', y='MonthlyIncome',hue='Attrition',legend= False, palette=['#5DCAA5','#D85A30'])
plt.title('Salary Distribution: Stayed vs Left')
plt.show()
```

## Attrition Rate by Job Role
Attrition rate broken down by job role, ranked from highest 
to lowest, to identify which roles are most at risk.

```python
plt.figure(figsize=(9,5))
Group_jobrole=df.groupby('JobRole')['AttritionFlag'].mean()
sort_jobrole=Group_jobrole.sort_values(ascending=False)
sort_jobrole.plot(kind='barh', color='#D85A30')
plt.title('Attrition Rate by Job Role')
plt.xlabel('Attrition Rate')
plt.tight_layout()
plt.show()
```

## Age Distribution by Attrition
Age distribution split by attrition status, showing how attrition 
is spread across different age groups.

```python
plt.figure(figsize=(8,4))
sns.histplot(data=df, x='Age', hue='Attrition', multiple='stack',
    palette=['#5DCAA5','#D85A30'], bins=20)
plt.title('Age Distribution by Attrition')
plt.show()
```
